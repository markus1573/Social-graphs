## E8 · PageRank by hand

Directed network on four nodes, $n=4$:

- $A \to B$, $A \to C$, $B \to C$, $C \to A$, $D \to B$

(D has no in-links. C points only at A.)

Use $\mathrm{PR}_i = \frac{1-\alpha}{n} + \alpha \sum_{j \to i} \mathrm{PR}_j / k_j^{\mathrm{out}}$ with $\alpha = 0.85$, and start from $\mathrm{PR} = (0.25, 0.25, 0.25, 0.25)$ in order $A, B, C, D$.

1. Apply the update **once**. Write the four new values (fractions or 3 decimals). Where does D's score come from?
2. Apply it a **second** time. Which node is ahead? Why can A score well with only one in-link?
3. Set $\alpha = 1$ (no teleport). What happens to D, and then to the walk? One sentence: what is teleportation *for*?
4. How does PageRank differ from eigenvector centrality in what one node's link is *worth* when that node has many out-links?


**Catch-up (not done live).** Closed-book takeaway: PageRank splits each page’s vote among its out-links, then mixes in a uniform teleport. Dangling pages and traps need that teleport; a node with no in-links gets *only* teleport.

**1.** $(1-\alpha)/n = 0.15/4 = 0.0375$. Out-degrees: $k_A^{\mathrm{out}}=2$, $k_B^{\mathrm{out}}=1$, $k_C^{\mathrm{out}}=1$, $k_D^{\mathrm{out}}=1$.

Who points at whom:
- $A \leftarrow C$ (C sends its whole score)
- $B \leftarrow A, D$ (A sends half; D sends all)
- $C \leftarrow A, B$ (A sends half; B sends all)
- $D \leftarrow$ nobody

First update from $(0.25,0.25,0.25,0.25)$:

- $\mathrm{PR}_A = 0.0375 + 0.85\cdot 0.25 = 0.250$
- $\mathrm{PR}_B = 0.0375 + 0.85\cdot(0.25/2 + 0.25) = 0.0375 + 0.85\cdot 0.375 = 0.356$
- $\mathrm{PR}_C = 0.0375 + 0.85\cdot(0.25/2 + 0.25) = 0.356$
- $\mathrm{PR}_D = 0.0375 + 0.85\cdot 0 = 0.038$

D’s score is **only teleport**. Nothing links to D, so the $\alpha$-sum is 0.

**2.** Second update from $(0.250,\,0.356,\,0.356,\,0.038)$:

- $\mathrm{PR}_A = 0.0375 + 0.85\cdot 0.356 = 0.340$
- $\mathrm{PR}_B = 0.0375 + 0.85\cdot(0.250/2 + 0.038) = 0.176$
- $\mathrm{PR}_C = 0.0375 + 0.85\cdot(0.250/2 + 0.356) = 0.447$
- $\mathrm{PR}_D = 0.038$

**C is ahead.** A has only one in-link, but it is **C**, who currently holds the most mass and sends *all* of it to A (C’s out-degree is 1). A half-vote from a nobody would not do that.

**3.** $\alpha=1$: $\mathrm{PR}_D=0$ immediately and stays 0 (no in-links, no teleport). Mass that started on D is dumped onto B in one step and then cycles in the strongly connected triple $A,B,C$. Teleportation is there so the walker cannot get stuck in a trap or a dangling page, and so every page (including sources like D) keeps a little mass.

**4.** Eigenvector centrality on $A$: a link from $j$ is worth $j$’s full score, no matter how many others $j$ also links to. PageRank: $j$ **splits** its score among its out-links. A hub that points at a thousand pages gives each of them almost nothing.


## E9 · Compared to what

1. Draw a network where a node of degree 2 has the **highest** betweenness. Draw a second network where the **highest-degree** node has betweenness 0.
2. Read this claim:

> "Node X has the highest betweenness in the network, so it is the critical broker. Its closeness of 0.55 means it can reach 55% of the nodes directly. That betweenness is far above what chance would produce."

  Mark each sentence *true*, *false*, or *not enough information*, and say which Essential it violates if it is false. (Assume you have **not** been shown a null-model table.)
3. Why is the **degree-preserving shuffle** the right null for "is this node's betweenness more than its degree?", and why is closeness usually *harder* to look surprising under that shuffle?


**Catch-up (not done live).** Closed-book takeaway: a centrality number is not a finding until you say which question it answers *and* compared to what. Highest betweenness can just be “this node has 106 links.” Closeness is inverse mean distance, not a percentage of the graph.

**1.** Degree 2, highest betweenness: two triangles (or two blobs) joined by a single bridge node, or a path $A-B-C-D$ — $B$ and $C$ have degree 2 and sit on many geodesics.

Highest degree, betweenness 0: a **clique** (everyone has the same high degree; every pair has a direct edge, so nobody sits on a shortest path). Same idea as “neighbours all linked to each other”: a hub whose neighbours form a clique among themselves has nothing to broker.

**2.** Three sentences:

- “X has the highest betweenness, so it is the critical broker.” **Not enough information** (and the “so” is the usual trap). Highest betweenness can be what the degree sequence already predicts (Spider-Man / degree-preserving shuffle). Violates *centrality, compared to what*.
- “Closeness 0.55 means it can reach 55% of the nodes directly.” **False.** $C^{\mathrm{close}}=(n-1)/\sum d_{ij}$ is the inverse of **mean distance** ($\approx 1.82$ steps here), not a coverage fraction. Violates the closeness definition.
- “That betweenness is far above chance.” **Not enough information** — no null was shown. Same Essential.

**3.** The degree-preserving shuffle keeps every $k_i$ and rewires the rest. If $i$’s betweenness in the real graph matches the shuffled graphs, the “broker” story is just “this node is a hub.” Hercules-style surprises are the ones with large $|z|$.

Closeness is hard to surprise under that shuffle because it is a **mean distance**, and in a small world distances are already short; the shuffle keeps the hubs, and hubs are close to everyone either way. Betweenness cares about *which* pairs you sit between, which rewiring can destroy.


## E10 · Mixing patterns

1. What is $r$ for a star (one hub, every other node linked only to it)? Is a **ring** assortative, disassortative, or neither — and why, from the definition of $r$?
2. Draw a 6-node network with $r > 0$ and one with $r < 0$. For each, sketch whether $k_{nn}(k)$ should rise, fall, or sit flat.
3. In one paragraph: what is *structural disassortativity*? Why must you compare Marvel's $r$ to the degree-preserving null before you call the universe "a technological network"?
4. Homophily (attribute assortativity) is tested by shuffling **labels** across nodes, not by rewiring links. Why is that the right null for "X-Men link to X-Men"?


**Catch-up (not done live).** Closed-book takeaway: $r$ is the correlation of the two degrees on a random *edge*. Stars are as disassortative as it gets. Heavy-tailed networks look disassortative even under a degree-preserving null — that is structure, not a social mechanism. Homophily tests freeze the links and shuffle the *labels*.

**1.** Star: every edge is hub–leaf $(k_{\mathrm{hub}}, 1)$. That is perfectly negative correlation: $r = -1$.

Ring: every node has degree 2, so both ends of every edge are $2$–$2$. There is **no degree variation** to correlate, so $r$ is undefined (Pearson needs a variance). The ring is **neither** assortative nor disassortative by degree.

**2.** $r>0$: two triangles (or two small cliques) joined by one edge — hubs sit next to hubs. $k_{nn}(k)$ **rises**.

$r<0$: a 6-node star (or a hub with five leaves). $k_{nn}(k)$ **falls** (the hub’s neighbours have degree 1; each leaf’s neighbour has high degree).

**3.** In a heavy-tailed network the hubs have so many links that they **must** attach mostly to small nodes — there are not enough other hubs to go around. That negative $r$ appears even if you randomly rewire while keeping the degrees (*structural disassortativity*). Marvel’s $r \approx -0.105$ vs shuffle $-0.118 \pm 0.009$ is that story, not “this is a technological network.” Compare to the null before naming the mechanism.

**4.** “X-Men link to X-Men” is a claim about **who has which label**, given the existing edges. Rewiring would smash the very structure you are asking about. Shuffle the team labels across nodes, keep every link: if the assortativity collapses, the original number was homophily; if it does not, it was just “there are many X-Men” or “X-Men are hubs.”


## E11 · Cliques

1. On **G**, list every **maximal** clique. What is the clique number?
2. A 6-clique has how many links? What is the smallest degree a node in that clique can have? Using only that, which nodes of G **cannot** sit in a 4-clique?
3. Why does clustering $C = 1$ mean the graph is a disjoint union of cliques, and why does $C = 0$ mean there is no 3-clique?


**Catch-up (not done live).** Closed-book takeaway: a $k$-clique has every pair linked, hence $k(k-1)/2$ edges, and every member has degree at least $k-1$. Maximal $\neq$ largest. $C=1$ means every wedge is a triangle (components are cliques); $C=0$ means no triangles.

**1.** Maximal cliques of **G** (cannot add a vertex without breaking completeness):

- $\{1,2,3\}$ (the only triangle)
- $\{2,4\}$, $\{4,5\}$, $\{3,5\}$, $\{4,6\}$, $\{6,7\}$

No third node is adjacent to both ends of those edges, so the 2-cliques are maximal. Clique number (size of the largest clique) = **3**.

**2.** A 6-clique has $6\cdot 5/2 = 15$ links. Each of its nodes is linked to the other five, so the smallest possible degree in that clique is **5**.

A 4-clique needs degree at least **3**. In G: $k_1=2$, $k_2=3$, $k_3=3$, $k_4=3$, $k_5=2$, $k_6=2$, $k_7=1$. So **1, 5, 6, 7** cannot sit in a 4-clique (without computing anything else). (In fact G has no 4-clique at all.)

**3.** Clustering $C$ is the fraction of wedges that are closed (triangles). $C=1$: every two neighbours of a node are themselves linked, so each connected component is a complete subgraph — a disjoint union of cliques. $C=0$: no wedge is closed, so there is no triangle, i.e. no 3-clique.
